## First overview

```python
check_df(df)
```

**Variables**
1. `Pregnancies` - number of times pregnant
2. `Glucose` - plasma glucose concentration (2h OGTT). <140 normal, 140-199 impaired, >=200 diabetes
3. `BloodPressure` - diastolic blood pressure (mm Hg)
4. `SkinThickness` - triceps skin fold thickness (mm)
5. `Insulin` - 2-hour post-prandial insulin (mu U/ml). <140 normal, 140-199 prediabetes, >=200 diabetes
6. `BMI` - body mass index
7. `DiabetesPedigreeFunction` - genetic influence proxy
8. `Age` (years)
9. `Outcome` - target (1 = diabetic, 0 = not diabetic)
